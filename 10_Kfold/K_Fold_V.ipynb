```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
```

```python
iris = load_iris()
```

Logistic regression

```python
lg_scores = cross_val_score(LogisticRegression(), iris.data, iris.target)
lg_scores
```

```python
np.average(lg_scores)
```

Decision Tree

```python
dt_scores = cross_val_score(DecisionTreeClassifier(), iris.data, iris.target)
dt_scores
```

```python
np.average(dt_scores)
```

svm

```python
svm_scores = cross_val_score(SVC(), iris.data, iris.target)
svm_scores
```

```python
np.average(svm_scores)
```

Random Forest

```python
rf_scores = cross_val_score(RandomForestClassifier(n_estimators=40), iris.data, iris.target)
rf_scores
```

```python
np.average(rf_scores)
```
